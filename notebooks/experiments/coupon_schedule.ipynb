# Coupon schedule

Xem lịch coupon dựng từ `term_sheet.csv`.

```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parents[1]
sys.path.insert(0, str(ROOT))
from src.bond_schedule import CouponSchedule, load_holiday_calendar

bond_df = pd.read_csv(ROOT / "datasets/raw/term_sheet.csv")
holiday_calendar = load_holiday_calendar(ROOT / "datasets/holiday")
schedule = CouponSchedule(bond_df, holiday_calendar, "vnd").build_coupon_schedule_df()
```

```python
cols = ["bond_id", "step", "start_date", "end_date", "accrual_end_date", "pay_date", "coupon_type", "fixed_rate"]

def show(bond_id):
    return schedule.loc[schedule["bond_id"] == bond_id, cols]
```
